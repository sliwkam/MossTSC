### **Parameter descriptions**

#### **discretization parameters**
- **`discretization_method`** (`"equal_width"` or `"equal_frequency"`):  
  determines how to discretize continuous values into bins:  
  - `"equal_width"`: divides data into bins of equal width
  - `"equal_frequency"`: ensures each bin contains approximately the same number of data points

- **`alpha`** (`5`):  
  the number of bins (or discrete symbols) used for discretization  

- **`beta`** (`4`):  
  defines the length of words (sequences of discretized symbols) or the dimensionality of the feature matrix  

- **`gamma`** (`7`):  
  the step size when creating words from discretized symbols. A larger step results in more spaced-out words  

---

### **statistical & matrix reduction parameters**
- **`statistical_reduction`** (`True`):  
  a boolean flag indicating whether to apply statistical reduction to the feature matrix  

- **`statistical_reduction_method`** (`"sum"`):  
  if `statistical_reduction` is `True`, this defines the method used for reduction. Possible methods:  
  - `"sum"`: computes the sum of values in each bin
  - `"std"`: uses standard deviation for reduction  

- **`matrix_reduction_method`** (`"pooling_std"`):  
  if `statistical_reduction` is `False`, this method is applied for matrix reduction. `"pooling_std"` suggests using a pooling mechanism based on standard deviation  

- **`standarization_method`** (`"ZScore"`):  
  the method used to standardize the data before processing. `"ZScore"` normalization transforms values to have a mean of 0 and a standard deviation of 1.  

- **`reduction_goal`** (`"2D"` or `"3D"`):  
  specifies the target output dimensionality after reduction  

---

### **experiment settings**
- **`seed`** (`0`):  
  determines the randomness in dataset splitting. If `0`, the original split is used without shuffling  

- **`batch_size`** (`16`):  
  the number of samples per batch during training  

- **`learning_rate`** (`0.1`):  
  the initial learning rate for the optimizer, controlling how much the model updates during training  

- **`epoch`** (`100`):  
  the maximum number of training epochs (full dataset passes)  

- **`patience`** (`25`):  
  the early stopping criterion: training will stop if no improvement is seen for this many epochs 

- **`device`** (`"cpu"`):  
  specifies the computation device (`"cpu"` or `"cuda"`) for running the model

## **Experiment output**

The `run_experiment` function outputs a list of dictionaries, where each dictionary contains details of a **single experiment run**. Each entry in the list represents a unique configuration of parameters and recorded performance metrics.


### **performance metrics (`metrics`)**
Each experiment records the following **evaluation metrics** on different datasets (`train`, `validation`, `test`):

- **`model`**: Neural network model used (`"AlexNet"`, `"LeNet"`, etc.).
- **`accuracy`**: Proportion of correctly classified samples.
- **`precision`**: Measure of correctness among predicted positive samples.
- **`recall`**: Measure of how well the model identifies positive samples.
- **`f1_score`**: Harmonic mean of precision and recall.
- **`log_loss`**: Logarithmic loss, which quantifies prediction uncertainty.
